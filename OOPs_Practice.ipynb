```python
class User:
    __password='Python@123'
    
    def change_password(self,old_password,new_password):
        if  old_password==self.__password:
            if len(new_password)>=8:
              self.__password=new_password
              return " New password set"
            else:
              return " New password contains minium 8 character"
        else:
            return " Enter right password"
    def check_password(self,password):
        if password==self.__password:
            return 'Correct password'
        else:
            return 'incorrect password'
       
        
```

```python
class ATM:
    __balance=5000
    __pin=5645
    def withdraw(self,amount,pin):
        if 0>=amount:
            return 'Amount must be greater than 0'
        
        elif pin==self.__pin:
            if amount<=self.__balance:
                self.__balance=self.__balance-amount
                return 'Transaction successful'
            else:
                return 'Insufficient amount'
        else:
            return 'wrong pin'
    def deposite(self,amount):
         if 0>=amount:
            return 'Amount must be greater than 0'
         else:
             self.__balance=self.__balance+amount
             return f'Current balance {self.__balance}'
    def check_pin(self,pin):
        if pin==self.__pin:
            return " Pin varify"
        else:
            return " Wrong pin"
    def get_balance(self,pin):
        if pin==self.__pin:
           return self.__balance
        else:
            return 'wrong pin'
        
            
            
```

# Inheritance

### Basic

```python
Problem 1: Single Inheritance
Concept: Parent class aur Child class.

Task:
1. Animal naam ki parent class banao.
2. Isme eat() method banao, jo "Animal is eating" print kare.
3. Dog naam ki child class banao, jo Animal ko inherit kare.
4. Dog class mein bark() method banao, jo "Dog is barking" print kare.
5. Dog class ka object banakar dono methods call karo.
```

```python
class Animal:
    def eat(self):
        print('Animal is eating')
class Dog(Animal):
    def bark(self):
        print('Dog is barking')
        
```

```python
dog=Dog()
#animal=Animal()
```

```python
dog.bark()
dog.eat()
```

```python
Problem 2: Inherited Attributes
Concept: Parent class ke attributes ko inherit karna.

Task:
1. Person naam ki class banao.
2. __init__() mein name aur age initialize karo.
3. Student naam ki child class banao.
4. Student ka name, age aur course store karo.
5. Student ki details display karne ke liye display() method banao.
Condition: Parent class ke name aur age ko child class mein dobara initialize karne ke bajaye super() ka use karo.
```

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
class Student(Person):
    def __init__(self,name,age,course):
        super().__init__(name,age)
        self.course=course
    def display(self):
        print(f'Name:{self.name}')
        print(f'Age:{self.age}')
        print(f'Course:{self.course}')
       

```

```python
std=Student('Rahid',33,'B.tech')
```

```python
std2=Student('Khan',55,'b.ed')
```

```python
std.display()
std2.display()
```

```python
Method Inheritance
Concept: Inherited methods aur child class ke apne methods.

Task:
1. Vehicle naam ki parent class banao.
2. Isme start() aur stop() methods banao.
3. Car naam ki child class banao.
4. Car mein drive() method banao.
5. Car ka object banakar teeno methods call karo.
```

```python
class Vehicle:
    def start(self):
        print('Vehicle is start')
    def stop(self):
        print('Vehicle is stop')
class Car(Vehicle):
    def drive(self):
        print('Car is driving')
```

```python
car=Car()
```

```python
car.start()
car.stop()
car.drive()
```

```python
Problem 4: Multilevel Inheritance
Concept: Ek class se doosri aur doosri se teesri class inherit karna.

Task:

Teen classes banao:
- Person: name attribute aur show_name() method.
- Employee: salary attribute aur show_salary() method.
- Manager: department attribute aur show_department() method.
```

```python
class Person:
    def __init__(self,name):
        self.name=name
    def show_name(self):
         print( f'Name:{self.name}')
class Employee(Person):
    def __init__(self,name,salary):
        super().__init__(name)
        self.salary=salary
    def show_salary(self):
        print(f'Salary:{self.salary}')
class Manager(Employee):
    def __init__(self,name,salary,department):
        super().__init__(name,salary)
        self.department=department
    def show_department(self):
        print( f'Department:{self.department}')
```

```python
man=Manager('Rahid',30000,'IT')
```

```python
man.show_name()
man.show_salary()
man.show_department()
```
