# Encapsulation


### Basic

```python
# Question 1
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def show_details(self):
        print(f" name is {self.name}")
        print(f" age is {self.age}")
```

```python
std=Student('Rahid',22)
std.show_details()
```

```python
# Question 2
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance
    
    def get_balance(self):
    
       return self.__balance
        
```

```python
acc=BankAccount(25000)
acc.get_balance()
```

```python
# Question 3
class Employee:
    def __init__(self,salary):
        self.__salary=salary
    def show_salary(self):
        return self.__salary
```

```python
es=Employee(25000)
print(es.show_salary()+5000)
```

```python
# Question 4
class User:
    
    def __init__(self,password):
        self.__password=password
    def check_password(self,password):
       return self.__password==password
```

```python
user=User(1238)
print(user.check_password(1236))
```

```python
# Question 5
class BankAccount:
    def __init__(self):
        
        self.__balance=1000
    def deposite(self,amount):
        self.amount=amount
        self.__balance=amount+self.__balance
        
        
    def get_balance(self):
        return self.__balance
    
    
```

```python
acc=BankAccount()
acc.deposite(500)
print(acc.get_balance())
```
