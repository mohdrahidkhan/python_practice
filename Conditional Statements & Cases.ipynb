# Conditional Statements

### Basic

```python
# Question 1 Positive,Negative or Zero
num=int(input())
if num>0:
    print("Number is positive")
elif num<0:
    print("Number is negative")
else:
    print("zero")
    
```

```python
# Q2. Even or Odd
num=int(input())
if num%2==0:
    print("Even")
else:
    print("Odd")
```

```python
#Q3. Voting Eligibility
age=int(input())
if age>=18:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
# Q4. Find the Greater Number
num1=int(input("Enter first no.: "))
num2=int(input("Enter first no.: "))
if num1>num2:
    print('First number grater than second number')
elif num2>num1:
    print('second number grater than First number')
else:
    print('Both are equal')
    
```

```python
# Q5. Pass or Fail
marks=int(input("Enter no. : "))
if  marks<0 or marks>100:
    print('Invalid marks')
elif marks>=33:
    print('Pass')
else:
    print('Fail')
```
