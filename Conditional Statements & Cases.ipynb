```python
# Q9. Electricity Bill Calculator
# Units	Rate
# First 100 units	₹5 per unit
# Next 100 units	₹7 per unit
# Above 200 units	₹10 per unit

units=int(input('Enter the units:'))

if units<0:
    print("Invalid units")
elif 0<=units<=100:
    print(f'Electricity bill:{units*5}')
elif units<=200:
    bill=(100*5)+((units-100)*7)
    print(f'Electricity bill:{bill}')
elif 201<=units:
    bill=(100*5)+(100*7)+((units-200)*10)
    print(f'Electricity bill:{bill}')
```

```python
# Q10. Triangle Validity and Type

# Step 1: Validity check

# Triangle valid tabhi hai jab:
# - a + b > c
# - a + c > b
# - b + c > a
# Aur teeno sides positive honi chahiye.

# Step 2: Triangle type determine karo
# - Teeno sides equal → Equilateral.
# - Koi do sides equal → Isosceles.
# - Teeno sides different → Scalene.
```

```python
a=int(input('Enter the side_a:'))
b=int(input('Enter the side_b:'))
c=int(input('Enter the side_c:'))
if a>0 and b>0 and c>0 and a + b > c and a + c > b and b + c > a:
    print('Triangle valid')
    if a==b and b==c and c==a:
        print('Equilateral')
    elif a==b or b==a or a==c:
        print('Isosceles')
    else:
        print('Scalene')
else:
    print('Invalid')
```

```python
# Q11. Income Tax Calculator
# Ek simplified income tax calculator banao.

# Annual taxable income input lo aur following slabs ke according tax calculate karo:
# Income Slab	Tax Rate
# ₹0–₹3,00,000	0%
# ₹3,00,001–₹7,00,000	5%
# ₹7,00,001–₹10,00,000	10%
# ₹10,00,001–₹12,00,000	15%
# ₹12,00,001–₹15,00,000	20%
# ₹15,00,000 se above	30%
```

```python
income=int(input('Enter the income:'))
if income<0:
    print('Invalid')
elif 0<=income<=300000:
    print(' your tax is: 0')
elif 300001<=income<=700000:
    tax=(income-300000)*5/100
    print(f' your tax is: {tax}')
elif 700001<=income<=1000000:
    tax=((income-700000)*10/100)+((700000-300000)*5/100)
    print(f' your tax is: {tax}')
elif 1000001<=income<=1200000:
    tax=((income-1000000)*15/100)+((1000000-700000)*10/100) +((700000-300000)*5/100)
    print(f' your tax is: {tax}')
elif 1200001<=income<=1500000:
    tax=((income-1200000)*20/100)+((1200000-1000000)*15/100) +((1000000-700000)*10/100) +((700000-300000)*5/100)
    print(f' your tax is: {tax}')
else:
    tax=((income-1500000)*30/100)+((1500000-1200000)*20/100)+((1200000-1000000)*15/100) +((1000000-700000)*10/100) +((700000-300000)*5/100)
    print(f' your tax is: {tax}')

```

```python
# Q12. ATM Withdrawal System
# Rules:
# 1. Amount positive hona chahiye.
# 2. Amount ₹100 ke multiple mein hona chahiye.
# 3. Balance sufficient hona chahiye.
# 4. Amount balance ke barabar ho sakta hai.
# 5. Transaction successful hone par balance update karo.
# 6. Invalid transaction mein balance change nahi hona chahiye.
```

```python
amount=int(input('Enter the amount:'))
balance=100000
if 0>=amount:
    print('Invalid')
elif amount%100==0:
    if amount<=balance:
        balance-=amount
        print('Transaction successfull')
        print(f'Remaining Balance :{balance}')
        
    else:
        print('Insufficient amount')
else:
    print('Amount must be multiple of 100')
    
```

```python
# Q13. Employee Bonus Eligibility
# Employee ki following details input lo:
# - Years of experience
# - Performance rating (1–5)
# - Annual salary
# Bonus eligibility ke rules:
# Condition	Bonus
# Experience ≥ 5 years and rating ≥ 4	Salary ka 20%
# Experience ≥ 3 years and rating ≥ 3	Salary ka 10%
# Baaki valid cases	No Bonus


# Additional rules:
# - Experience negative nahi ho sakta.
# - Rating 1–5 ke beech honi chahiye.
# - Salary positive honi chahiye.
```

```python
experience=int(input('Enter Years of experience:'))
rating=int(input('Enter Performance rating (1–5): '))
salary=int(input('Annual salary:'))
if 0>=rating or rating>5:
    print('wrong rating')
elif 0>=experience:
    print('Wrong experience')
elif salary<=0:
    print('wrong salary')
elif experience >= 5  and rating >= 4:
    print(f'You bonus is 20% of salary: {salary*20/100}')
elif experience >= 3  and rating >= 3:
    print(f'You bonus is 10% of salary: {salary*10/100}')
else:
    print('No bonus')
```

```python
# Q14. Number Classification System
# Ek integer input lo aur uski multiple properties classify karo.

# Program ko determine karna hai:
# 1. Positive, negative ya zero.
# 2. Even ya odd.
# 3. Agar positive hai, toh 10 se chhota, 10–100 ke beech ya 100 se bada.
# 4. Agar number 5 aur 7 dono se divisible hai, toh special message print karo.
# Important: Zero ko even treat karna hai, lekin uske liye positive-number wali range classification apply nahi hogi.
```

```python
num=int(input('Enter the number:'))
if num>0:
    print("Positive")
    if num<10:
       print('Number smaller then 10')  
    elif 10<=num<=100:
       print('Number in range of 10-100')       
    else:
       print('Number bigger then 100')
elif num<0:
    print("Negative")
else:
    print("Zero")
    
if num%2==0:
    print('Even')
else:
    print("Odd")

if num%5==0 and num%7==0:
    print('Divisible by both 5 and 7')
    
```

```python
# Q15. E-commerce Discount Engine
# Ek shopping application ke liye discount calculator banao.

# User se input lo:
# - Cart amount
# - Customer type (regular ya premium)
# - Membership status (yes ya no)
# Discount rules:
# Condition	Discount
# Cart amount ₹5,000 se kam	0%
# Cart amount ₹5,000–₹9,999, regular customer	5%
# Cart amount ₹5,000–₹9,999, premium customer	10%
# Cart amount ₹10,000 ya zyada, regular customer	10%
# Cart amount ₹10,000 ya zyada, premium customer	20%


# Additional rule: Agar customer ke paas valid membership hai, toh calculated discount ke baad ₹500 ka additional discount milega.

# Important assumptions:
# - Cart amount negative nahi ho sakta.
# - Customer type sirf regular ya premium ho sakta hai.
# - Membership status sirf yes ya no ho sakta hai.
# - Final payable amount negative nahi hona chahiye.
```

```python
amount=int(input('Enter Cart amount: '))
type=input('Enter Customer type (regular ya premium): ')
status=input('Enter Membership status (yes ya no): ')
if amount<0:
    print('Enter right amount')
elif type!='regular' and type!='premium':
    print('Enter right Customer type (regular ya premium)')
elif status!='yes' and status!='no':
    print('Enter right Membership status (yes ya no)')
else:
    if amount<5000:
       discount=0
       print(f'0% Discount is {discount}')
    elif 5000<=amount<=9999:
       discount=amount*5/100
       print(f'5% Discount is {discount}')
    elif 10000<=amount and type=='premium':
       discount=amount*20/100
       print(f'20% Discount is {discount}')
    else:
       discount=amount*10/100
       print(f'10% Discount is {discount}')
    if status=='yes':
       total_discount=discount+500
       total_amount=amount-total_discount
       if total_amount<0:
          print(f'total discount:{total_discount}')
          print('Total payable bill:0')
       else:
          print(f'total discount:{total_discount}')
          print(f'Total payable bill:{total_amount}')
    else:        
       print(f'Total payable bill:{amount-discount}')

     
```
